```python
import pandas as pd
import matplotlib.pyplot as plt

#Read in file from Drive
df = pd.read_csv('/content/drive/MyDrive/diamonds.csv', index_col=0)
sp = '\n\n'

#Print first seven rows
print(df.head(7),sp)

#Print last seven rows
print(df.tail(7),sp)

#describe numerical data
print(df.describe(),sp)

#describe textual data
print(df['clarity'].describe(),sp)
print(df['cut'].describe(),sp)
print(df['color'].describe(),sp)

#unique category values
print(df.cut.unique(),sp)
print(df.color.unique(),sp)
print(df.clarity.unique(),sp)

print(df.carat.hist(),sp)
plt.title('Carat')
plt.show()
print(df.price.hist(),sp)
plt.title('Price')
plt.show()
print(df.depth.hist(),sp)
plt.title('Depth')
plt.show()
print(df.table.hist(),sp)
plt.title('Table')
plt.show()
print(df.x.hist(),sp)
plt.title('X')
plt.show()
print(df.y.hist(),sp)
plt.title('Y')
plt.show()
print(df.z.hist(),sp)
plt.title('Z')
plt.show()



```
